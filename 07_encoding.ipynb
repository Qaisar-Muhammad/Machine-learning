```python
# import linbraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
#dataset
df = pd.read_csv('tip.csv')
df.head()
```

```python
df['time'].value_counts()
```

```python
# let's encode 'time' column using Label Encoding with sklearn
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
le = LabelEncoder()
df['encoded_time'] = le.fit_transform(df['time']) 
df.head()
```

```python
df['encoded_time'].value_counts()
```

```python
df['day'].value_counts()
```

```python
#ordinal encoding on 'day' column using OrdinalEncoder
ordinal_enc = OrdinalEncoder(categories=[['Thur', 'Fri', 'Sat', 'Sun']])
df['encoded_day'] = ordinal_enc.fit_transform(df[['day']])
df.head()
```

```python
df['encoded_day'].value_counts()
```

```python
# one hot encoding on 'sex' column
ohe = OneHotEncoder()
ohe.fit_transform(df[['sex']]).toarray()
```
