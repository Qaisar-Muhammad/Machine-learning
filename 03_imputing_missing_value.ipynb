```python
from sklearn.preprocessing import LabelEncoder
coloum_to_encode=['sex','embarked','who','deck','class','embark_town','alive']
label_encoders={}
for col in coloum_to_encode:
    le=LabelEncoder()
    df[col]=le.fit_transform(df[col])
    label_encoders[col]=le
df.head()
```

```python
imputer =IterativeImputer(max_iter=10,random_state=0)  # iterative imputer class called with max iteration 10 and random state 0
columns_to_impute=['age','embarked','deck','embark_town']
for col in columns_to_impute:
    df[col]=imputer.fit_transform(df[[col]])
df.isnull().sum().sort_values(ascending=False)
```
