```python
# pandas task 
# show 5 cars with highest and 5 with lowest mpg
#filter cars that have 6 or 8 cylinders and hp >150
#create  new column : power to weight ratio = hp/weight
#compute mean mpg grouped by number of cylinders(groupedby)

import pandas as pd

# Manually create dataset
data = {
    'car': ['Ford', 'BMW', 'Toyota', 'Honda', 'Audi', 'Chevy', 'Nissan', 'Kia'],
    'mpg': [18, 25, 30, 32, 22, 15, 28, 35],
    'cylinders': [8, 6, 4, 4, 6, 8, 4, 4],
    'hp': [200, 180, 140, 130, 170, 220, 150, 120],
    'weight': [3500, 3200, 2800, 2700, 3000, 3800, 2900, 2600]
}

df = pd.DataFrame(data)

# 1. Show 5 cars with highest mpg
print("5 cars with highest mpg:")
print(df.sort_values(by='mpg', ascending=False).head(5))

# 2. Show 5 cars with lowest mpg
print("\n5 cars with lowest mpg:")
print(df.sort_values(by='mpg').head(5))

# 3. Filter cars with 6 or 8 cylinders and hp > 150
filtered_cars = df[(df['cylinders'].isin([6, 8])) & (df['hp'] > 150)]
print("\nCars with 6 or 8 cylinders and hp > 150:")
print(filtered_cars)

# 4. Create new column: power to weight ratio
df['power_to_weight'] = df['hp'] / df['weight']
print("\nDataset with power to weight ratio:")
print(df)

# 5. Mean mpg grouped by number of cylinders
mean_mpg = df.groupby('cylinders')['mpg'].mean()
print("\nMean mpg grouped by cylinders:")
print(mean_mpg)

```

```python
# average mpg,hp,wt,grouped by cyl and gear
# create pivot table :mean mpg by cyl(row) vs gear(coloumn)
#find car with best hp /mpg raio (lowest value =most effient power usage )
import pandas as pd

# Manually create dataset
data = {
    'car': ['Ford', 'BMW', 'Toyota', 'Honda', 'Audi', 'Chevy', 'Nissan', 'Kia'],
    'mpg': [18, 25, 30, 32, 22, 15, 28, 35],
    'cyl': [8, 6, 4, 4, 6, 8, 4, 4],
    'hp': [200, 180, 140, 130, 170, 220, 150, 120],
    'wt': [3.5, 3.2, 2.8, 2.7, 3.0, 3.8, 2.9, 2.6],
    'gear': [3, 5, 5, 5, 4, 3, 5, 5]
}

df = pd.DataFrame(data)

# -----------------------------------
# 1️⃣ Average mpg, hp, wt grouped by cyl and gear
avg_values = df.groupby(['cyl', 'gear'])[['mpg', 'hp', 'wt']].mean()
print("Average mpg, hp, wt grouped by cyl and gear:")
print(avg_values)

# -----------------------------------
# 2️⃣ Pivot table: mean mpg by cyl (rows) vs gear (columns)
pivot_mpg = pd.pivot_table(
    df,
    values='mpg',
    index='cyl',
    columns='gear',
    aggfunc='mean'
)
print("\nPivot table (mean mpg by cyl vs gear):")
print(pivot_mpg)

# -----------------------------------
# 3️⃣ Find car with best hp/mpg ratio (lowest = most efficient)
df['hp_mpg_ratio'] = df['hp'] / df['mpg']

best_car = df.loc[df['hp_mpg_ratio'].idxmin()]
print("\nCar with best hp/mpg ratio (most efficient):")
print(best_car)
```

```python
# draw mpg vs hp colored by wt 
# bar plot -average mpg  by cylinder count 
# boxplot of mpg by transmission (am:0=auto,1=manual)
# scatter mpg vs qsec + linear trend  line (np.polyfit degree 1)
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = {
    'car': ['Ford', 'BMW', 'Toyota', 'Honda', 'Audi', 'Chevy', 'Nissan', 'Kia'],
    'mpg': [18, 25, 30, 32, 22, 15, 28, 35],
    'cyl': [8, 6, 4, 4, 6, 8, 4, 4],
    'hp': [200, 180, 140, 130, 170, 220, 150, 120],
    'wt': [3.5, 3.2, 2.8, 2.7, 3.0, 3.8, 2.9, 2.6],
    'am': [0, 1, 1, 1, 0, 0, 1, 1],
    'qsec': [18.5, 17.8, 16.5, 16.2, 17.9, 19.0, 16.8, 15.9]
}

df = pd.DataFrame(data)

# 1 mpg vs hp colored by wt
plt.figure()
plt.scatter(df['hp'], df['mpg'], c=df['wt'])
plt.xlabel('hp')
plt.ylabel('mpg')
plt.title('mpg vs hp colored by wt')
plt.show()

# 2 bar plot: average mpg by cylinder count
avg_mpg = df.groupby('cyl')['mpg'].mean()
plt.figure()
plt.bar(avg_mpg.index, avg_mpg.values)
plt.xlabel('cyl')
plt.ylabel('average mpg')
plt.title('Average mpg by cylinder count')
plt.show()

# 3 boxplot of mpg by transmission
plt.figure()
df.boxplot(column='mpg', by='am')
plt.xlabel('am (0=auto, 1=manual)')
plt.ylabel('mpg')
plt.title('mpg by transmission')
plt.suptitle('')
plt.show()

# 4 scatter mpg vs qsec with linear trend line
x = df['qsec']
y = df['mpg']
m, b = np.polyfit(x, y, 1)

plt.figure()
plt.scatter(x, y)
plt.plot(x, m*x + b)
plt.xlabel('qsec')
plt.ylabel('mpg')
plt.title('mpg vs qsec with linear trend')
plt.show()
```

```python
# EDA
# create two new coloumn:
# km_per_liter

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 0: Manually create dataset (simplified mtcars style)
data = {
    'car': ['Mazda RX4', 'Honda Civic', 'Toyota Corolla', 'Ford Mustang', 'Chevy Camaro', 'Datsun 710', 'Merc 240D', 'Ferrari Dino'],
    'mpg': [21, 30, 33, 18, 15, 22, 24, 19],
    'cyl': [6, 4, 4, 8, 8, 4, 4, 6],
    'hp': [110, 75, 68, 150, 175, 97, 62, 175],
    'wt': [2.62, 1.8, 1.7, 3.2, 3.4, 2.32, 2.2, 2.77],
    'qsec': [16.46, 18.5, 19.2, 17.0, 16.8, 18.9, 20.0, 15.5],
    'am': [1, 1, 1, 0, 0, 1, 0, 1]
}

df = pd.DataFrame(data)

# Step 1: Create new columns
df['km_per_liter'] = df['mpg'] * 0.425144
df['performance_index'] = df['hp'] / df['wt']

# Step 2a: Which cylinder count has highest average performance_index?
avg_perf = df.groupby('cyl')['performance_index'].mean()
print("Average performance_index by cylinder:")
print(avg_perf)

plt.figure()
avg_perf.plot(kind='bar', color='skyblue')
plt.xlabel('Cylinders')
plt.ylabel('Average Performance Index')
plt.title('Average Performance Index by Cylinder Count')
plt.savefig('figure1.png')
plt.show()

# Step 2b: Relationship between weight and acceleration (qsec)
plt.figure()
sns.scatterplot(data=df, x='wt', y='qsec')
plt.xlabel('Weight (1000 lbs)')
plt.ylabel('qsec (1/4 mile time)')
plt.title('Weight vs Acceleration')
plt.savefig('figure2.png')
plt.show()

# Step 2c: Compare automatic vs manual transmission on mpg and hp
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
sns.boxplot(data=df, x='am', y='mpg')
plt.xlabel('Transmission (0=Auto, 1=Manual)')
plt.title('MPG by Transmission')

plt.subplot(1,2,2)
sns.boxplot(data=df, x='am', y='hp')
plt.xlabel('Transmission (0=Auto, 1=Manual)')
plt.title('HP by Transmission')
plt.tight_layout()
plt.savefig('figure3.png')
plt.show()

# Step 2d: Correlation heatmap of numeric columns
plt.figure()
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.savefig('figure4.png')
plt.show()
```
